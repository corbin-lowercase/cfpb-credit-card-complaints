```python
"""Load multiple csv files and check for cohesion.

Files should have same number of columns, same data types, and no duplicate entries.
"""
# TODO: move this code to a separate load_to_sql.py file


from pathlib import Path
import pandas as pd


dir = Path("../data/raw")

# Locate all files to be loaded.
files = sorted(dir.glob("*.csv"))
print(f"Found {len(files)} files: ", [f.name for f in files])

dataframes = []

for f in files:
    df = pd.read_csv(f, dtype=str)  # Maintain type as str to preserve formatting.
    df["source_file"] = f.name  # Add column
    print(f"{f.name}: {len(df):,} rows, {df.shape[1]} columns")
    dataframes.append(df)

# Check that all files have same columns.
first_cols = list(dataframes[0].columns)
for f, df in zip(files, dataframes):
    if list(df.columns) != first_cols:
        print(f"Column mismatch in {f.name}")

complaints = pd.concat(dataframes, ignore_index=True)

# Check that there is no overlap between files.
duplicates = complaints["Complaint ID"].duplicated(keep=False)
print(f"Total rows: {len(complaints):,}")
print(f"Rows with a duplicated Complaint ID: {duplicates.sum():,}")

df.tail()
```
